### Q4

```python
with open('numbers.txt','w') as c:

    i=1
    for j in range(5):
        c.write(f"{i} \n")
        i+=1
    

```

### Q5

### Q8 (Medium)
Ek file `sample.txt` se read karke uski saari lines ko ek Python list mein convert karein (`readlines()` ka use karke).

```python
with open('sample.txt','r') as s:
    addlist=[]
    for line in s.readlines():
      addlist.append(line.strip())# strip will be remove the /n
    print(addlist)
```

### Q17

```python
try:
    number1 = int(input("Enter Number 1"))
    number2 = int(input("Enter Number 2"))
    number1/number2
except ValueError:
    print("wrong input, please enter NUMBER value")
except ZeroDivisionError:
    print("You can not divide a number by 0")
```

### Q18
`else` block ka istemal `try-except` ke sath karein jo sirf tab chale jab koi error na aaye.

```python
try:
    num = int(input("Please enter Number"))
    result = 20 / num
except ZeroDivisionError:
    print("You can not divide by 0")
else:
    print("No Error")
    print(f"Result is {result}")
```

### Q19 
`raise` keyword ka use karke age validation ka function banayein. Agar age 18 se kam ho toh custom `ValueError("Age must be 18 or older")` raise karein.

```python
def date_check(age):
    if age<18:
        raise ValueError("Age must be 18 or older")
    else:
        print("Age is Valid")

date_check(16)
```
