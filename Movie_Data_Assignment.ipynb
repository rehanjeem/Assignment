### **Question 27**
(Visualization) Har saal (Released_Year) kitni movies release hueen, iska ek Line Plot banayein.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Year column ko numeric format mein convert karein
data['Released_Year'] = pd.to_numeric(
    data['Released_Year'], errors='coerce'
)

# Missing years remove karke har saal movies count karein
yearly_movies = (
    data.dropna(subset=['Released_Year'])
        .groupby('Released_Year')
        .size()
        .sort_index()
)

# Line Plot
plt.figure(figsize=(12, 6))

sns.lineplot(
    x=yearly_movies.index,
    y=yearly_movies.values,
    color='royalblue',
    linewidth=2,
    marker='o',
    markersize=4
)

plt.title('Number of Movies Released Each Year',
          fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Release Year', fontsize=12)
plt.ylabel('Number of Movies', fontsize=12)

plt.grid(True, linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
```

### **Question 28**
(Visualization) 'No_of_Votes' ke mutabiq top 5 highest voted movies ka ek Horizontal Bar Chart banayein.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Votes ko numeric format mein convert karein
data['No_of_Votes'] = pd.to_numeric(
    data['No_of_Votes'].astype(str).str.replace(',', '', regex=False),
    errors='coerce'
)

# Top 5 highest-voted movies
top_movies = (
    data.dropna(subset=['Series_Title', 'No_of_Votes'])
    .nlargest(5, 'No_of_Votes')
    .sort_values('No_of_Votes', ascending=True)
)

# Horizontal Bar Chart
plt.figure(figsize=(11, 6))

sns.barplot(
    data=top_movies,
    x='No_of_Votes',
    y='Series_Title',
    hue='Series_Title',
    palette='viridis',
    legend=False
)

plt.title('Top 5 Movies by Number of Votes',
          fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Number of Votes', fontsize=12)
plt.ylabel('Movie Title', fontsize=12)

# Har bar par vote count show karein
for i, votes in enumerate(top_movies['No_of_Votes']):
    plt.text(votes, i, f' {votes:,.0f}', va='center', fontsize=10)

plt.grid(axis='x', linestyle='--', alpha=0.3)
sns.despine()
plt.tight_layout()
plt.show()
```

### **Question 36**
(Visualization) 'Drama' aur 'Action' genres ki movies ke 'Gross' revenue ko compare karne ke liye ek Grouped Box Plot ya Violin Plot banayein.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Gross revenue ko numeric format mein convert karein
data['Gross'] = pd.to_numeric(
    data['Gross'].astype(str).str.replace(',', '', regex=False),
    errors='coerce'
)

# Drama aur Action movies filter karein
plot_data = data[
    data['Genre'].str.contains(r'Drama|Action', na=False)
].copy()

# Har movie ko relevant genre assign karein
plot_data['Genre_Group'] = plot_data['Genre'].apply(
    lambda g: 'Drama' if 'Drama' in g.split(', ')
    else 'Action' if 'Action' in g.split(', ')
    else None
)

# Missing values remove karein
plot_data = plot_data.dropna(subset=['Gross', 'Genre_Group'])

# Box Plot
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=plot_data,
    x='Genre_Group',
    y='Gross',
    hue='Genre_Group',
    palette={'Drama': 'cornflowerblue', 'Action': 'coral'},
    legend=False
)

plt.title('Gross Revenue: Drama vs Action Movies',
          fontsize=15, fontweight='bold')
plt.xlabel('Genre', fontsize=12)
plt.ylabel('Gross Revenue', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()
```

### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Numeric columns convert karein
data['IMDB_Rating'] = pd.to_numeric(
    data['IMDB_Rating'], errors='coerce'
)
data['Meta_score'] = pd.to_numeric(
    data['Meta_score'], errors='coerce'
)

# Top 5 highest-rated movies select karein
top5 = (
    data.dropna(subset=['Series_Title', 'IMDB_Rating', 'Meta_score'])
    .nlargest(5, 'IMDB_Rating')
    .copy()
)

# IMDb rating ko 100 ke scale par convert karein
top5['IMDB_Rating_Scaled'] = top5['IMDB_Rating'] * 10

# Data ko comparison ke liye reshape karein
plot_data = top5.melt(
    id_vars='Series_Title',
    value_vars=['Meta_score', 'IMDB_Rating_Scaled'],
    var_name='Score_Type',
    value_name='Score'
)

# Side-by-Side Bar Chart
plt.figure(figsize=(13, 7))

sns.barplot(
    data=plot_data,
    x='Series_Title',
    y='Score',
    hue='Score_Type',
    palette=['#3498DB', '#F39C12']
)

plt.title(
    'Top 5 Highest-Rated Movies: Meta Score vs IMDb Rating',
    fontsize=15,
    fontweight='bold',
    pad=15
)
plt.xlabel('Movie Title', fontsize=12)
plt.ylabel('Score (Out of 100)', fontsize=12)
plt.ylim(0, 105)
plt.xticks(rotation=25, ha='right')
plt.legend(title='Score Type', labels=['Meta Score', 'IMDb Rating × 10'])
plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()
```
